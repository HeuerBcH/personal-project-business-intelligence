```python
import pandas as pd
import numpy as np
```

```python
df_matches = pd.read_excel('../data/raw/WorldCupMatches.xlsx')
df_matches.head()
```

```python
df_matches.drop(columns=["Win conditions"], inplace=True)
df_matches.head()
```

```python
df_matches["Datetime"] = df_matches["Datetime"].str.strip()

df_matches["Datetime"] = pd.to_datetime(
    df_matches["Datetime"], 
    format='%d %b %Y - %H:%M', 
    errors='coerce'
)
```

```python
df_matches.info()
```

```python
df_matches.to_csv('../data/processed/WorldCupMatches.csv', index=False)
```

```python
df_worldcups = pd.read_excel('../data/raw/WorldCups.xlsx')
df_worldcups.head()
```

```python
df_worldcups.info()
```

```python
df_worldcups.loc[df_worldcups['Attendance'] == 363, 'Attendance'] = 363000
df_worldcups.loc[df_worldcups['Attendance'] == 375.7, 'Attendance'] = 375700
df_worldcups.head(10)
```

```python
import pandas as pd

df_worldcups['Attendance'] = df_worldcups['Attendance'].astype(str)

df_worldcups['Attendance'] = df_worldcups['Attendance'].str.replace('.', '', regex=False)

df_worldcups['Attendance'] = df_worldcups['Attendance'].str.replace(',', '.', regex=False)

df_worldcups['Attendance'] = pd.to_numeric(df_worldcups['Attendance'], errors='coerce')

df_worldcups.head(10)
```

```python
df_worldcups.to_csv('../data/processed/WorldCupsData.csv', index=False)
```
